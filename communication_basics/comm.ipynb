```python
import time

with serial.Serial(motors.PORT, motors.BAUD, timeout=0.05) as ser:
    try:
        while True:
            positions = [motors.read_position(ser, i) for i in range(1, 7)]
            print(positions, end="\r")        
            time.sleep(0.1)
    except KeyboardInterrupt:       
        print("\nstopped")
```
